```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
import pandas as pd
from sklearn.model_selection import cross_val_score

# 读取数据
data = pd.read_excel("/Users/macbook/folder/data/snvFL199.xlsx")
X = data.values[:, 1:]
y = data.values[:, 0]
print(X.shape)

# 绘制原始光谱图
plt.figure(figsize=(8, 6))
plt.plot(np.linspace(942, 1701, X.shape[1]), np.transpose(X))
fonts = 20
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

# Customize y-axis tick intervals and decimal places
plt.gca().get_yaxis().set_major_locator(plt.MaxNLocator(integer=False))  # To allow non-integer tick intervals
plt.gca().yaxis.set_major_formatter(plt.FormatStrFormatter('%.2f'))  # Format tick labels to two decimal places

plt.xlabel('波长(nm)', fontsize=fonts)
plt.ylabel('反射率', fontsize=fonts)
plt.tight_layout()
plt.grid(False)

# 在最左上角添加 "(a)"
plt.text(915, max(np.max(X, axis=0))*1.005 , "(a)", fontsize=16, color='black', va='top', ha='left')
plt.gcf().tight_layout(pad=0,rect=(0.01,0.01,0.95,0.98))
# 保存为TIFF格式
plt.savefig('/Users/macbook/Desktop/论文原图/a.tif', format='tiff', dpi=300, bbox_inches='tight')
# 保存为JPG格式
plt.savefig('/Users/macbook/Desktop/论文原图/a.jpg', format='jpg', dpi=300, bbox_inches='tight')

plt.show() 
```
