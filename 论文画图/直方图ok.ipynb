```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
with plt.style.context(('seaborn-whitegrid')):  
    plt.rcParams['font.sans-serif'] = ['Times New Roman']
    plt.rcParams['axes.unicode_minus'] = False

    df = pd.read_excel(r'/Users/macbook/Documents/data/199个样本画直方图和箱线图.xlsx',header=None)#读取数据

    fig,ax = plt.subplots(figsize=(6,4),facecolor='white')#创建图形
    sns.histplot(df.values,bins=24,ax=ax,edgecolor='w',kde=True,label=None) #直方图
    plt.grid(False) 
    #ax.bar_label(ax.containers[0],labels=np.histogram(df.values,bins=15)[0]) #标注数据
    plt.xticks(fontsize=15)  
    plt.yticks(fontsize=15)  
    ax.set_xlabel('Leaf nitrogen content(%)',fontsize=18) #标题
    ax.set_ylabel('Frequency',fontsize=18)
    ax.legend_.remove() #去除图例
    ax.set_xticks(np.linspace(2.1,4.7,14)) #设置横轴刻度
    # 将横轴刻度中的数字保留两位小数  
    xtick_labels = [np.round(tick, 1) for tick in ax.get_xticks()]  
    ax.set_xticklabels(xtick_labels)  
    # plt.savefig('fig.jpg',dpi=600,bbox_inches='tight')
    plt.show()
```
