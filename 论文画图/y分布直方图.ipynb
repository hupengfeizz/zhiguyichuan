```python
import pandas as pd  
import matplotlib.pyplot as plt  
import numpy as np
# 读取Excel文件  
df = pd.read_excel('/Users/macbook/Documents/data/FL200_N.xlsx', engine='openpyxl')  
  
# 假设我们要读取的列名是'Column1'  
data = df['N%']  
  
# 创建直方图，通过调整bins参数的width来增加箱子之间的间距  
bins = np.arange(min(data), max(data) + 1.5, 0.5)  # 修改width参数来增加间距  
plt.hist(data, bins=13, edgecolor='black', histtype='bar', label='Frequency')  
  
# 在每个箱子的顶部添加标签  
for rect in plt.gca().get_children():  
    if 'bar' in str(type(rect)):  
        plt.text(rect.get_x() + rect.get_width() / 2, rect.get_y() + rect.get_height() / 2, str(int(rect.get_width())), ha='center')  
  
# 设置图表标题和坐标轴标签  
plt.title('Distribution of Column1')  
plt.xlabel('Column1')  
plt.ylabel('Frequency')  
  
# 显示图例，并将图例放在右下角  
plt.legend(loc='upper right')  
  
# 显示直方图  
plt.show()
```

```python
import pandas as pd  
import matplotlib.pyplot as plt  
import numpy as np  
  
# 读取Excel文件  
df = pd.read_excel('/Users/macbook/Documents/data/FL200_N.xlsx', engine='openpyxl')  
  
# 假设我们要读取的列名是'Column1'  
data = df['N%']  
  
# 修改数据范围以适应12个箱子  
data_min = 2  
data_max = 4.5  
bin_width = (data_max - data_min) / 12  # 计算每个箱子的宽度  
bin_edges = np.arange(data_min, data_max + bin_width, bin_width)  # 创建箱子边界  
bins = bin_edges[:-1]  # 移除最后一个箱子边界，因为最后一个箱子到最大值包括最大值本身  
  
# 创建直方图，设置箱子数量为12  
plt.hist(data, bins=bins, edgecolor='black', histtype='bar', label='Frequency', rwidth=0.8)  
  
# 在每个箱子的顶部添加标签  
for rect in plt.gca().get_children():  
    if 'bar' in str(type(rect)):  
        bin_center = rect.get_x() + rect.get_width() / 2  
        bin_height = rect.get_height()  # This represents the frequency  
        plt.text(bin_center, bin_height + 1, str(int(bin_height)), ha='center')  # Adding 1 to the height to place the text above the bin  
  
# 设置图表标题和坐标轴标签  
plt.title('Distribution of Column1')  
plt.xlabel('Column1')  
plt.ylabel('Frequency')  
  
# 显示图例，并将图例放在右下角  
plt.legend(loc='upper right')  
  
# 显示直方图  
plt.show()
```

```python
import pandas as pd  
import matplotlib.pyplot as plt  
import numpy as np  
from matplotlib.offsetbox import AnchoredText  
  
# 读取Excel文件  
df = pd.read_excel('/Users/macbook/Documents/data/FL200_N.xlsx', engine='openpyxl')  
  
# 假设我们要读取的列名是'Column1'  
data = df['N%']  
  
# 修改数据范围以适应12个箱子  
data_min = 2  
data_max = 4.5  
bin_width = (data_max - data_min) / 12  # 计算每个箱子的宽度  
bin_edges = np.arange(data_min, data_max + bin_width, bin_width)  # 创建箱子边界  
bins = bin_edges[:-1]  # 移除最后一个箱子边界，因为最后一个箱子到最大值包括最大值本身  
  
# 创建直方图，设置箱子数量为12  
hist_result = plt.hist(data, bins=bins, edgecolor='black', histtype='bar', label='Frequency', rwidth=0.8)  
  
# 在每个箱子的顶部添加频率标签  
for bin_index, bin_height in enumerate(hist_result[0]):  
    if bin_index == 0:  # 处理第一个箱子，因为它没有单独的标签  
        anchored_text = AnchoredText(f"{bin_index + 1}: {bin_height}", loc=8, prop={'size': 8})  
        plt.gca().add_artist(anchored_text)  
    else:  
        anchored_text = AnchoredText(f"{bin_index}: {bin_height}", loc=8, prop={'size': 8})  
        plt.gca().add_artist(anchored_text)  
  
# 设置图表标题和坐标轴标签  
plt.title('Distribution of Column1')  
plt.xlabel('Column1')  
plt.ylabel('Frequency')  
  
# 显示图例，并将图例放在右下角  
plt.legend(loc='upper right')  
  
# 显示直方图  
plt.show()
```

```python
import pandas as pd  
import matplotlib.pyplot as plt  
import numpy as np  
  
# 读取Excel文件  
df = pd.read_excel('/Users/macbook/Documents/data/FL200_N.xlsx', engine='openpyxl')  
  
# 假设我们要读取的列名是'Column1'  
data = df['N%']  
  
# 修改数据范围以适应12个箱子  
data_min = 2  
data_max = 4.5  
bin_width = (data_max - data_min) / 12  # 计算每个箱子的宽度  
bin_edges = np.arange(data_min, data_max + bin_width, bin_width)  # 创建箱子边界  
bins = bin_edges[:-1]  # 移除最后一个箱子边界，因为最后一个箱子到最大值包括最大值本身  
  
# 创建直方图，设置箱子数量为12  
hist_result = plt.hist(data, bins=bins, edgecolor='black', histtype='bar', label='Frequency', rwidth=0.8)  
  
# 在每个箱子的顶部添加标签  
for bin_index, bin_height in enumerate(hist_result[0]):  
    bin_center = bins[bin_index] + bin_width / 2  # 计算箱子的中心位置  
    bin_frequency = len(df[df['N%'] == bins[bin_index]])  # 计算该箱子代表的数据在所有数据中出现的频率  
    plt.text(bin_center, bin_height + 1, str(bin_frequency), ha='center')  # 在箱子的中心位置添加标签，并显示频率

  
# 设置图表标题和坐标轴标签  
plt.title('Distribution of Column1')  
plt.xlabel('Column1')  
plt.ylabel('Frequency')  
  
# 显示图例，并将图例放在右下角  
plt.legend(loc='upper right')  
  
# 显示直方图  
plt.show()
```

```python
import matplotlib.pyplot as plt  
import numpy as np  
  
# 读取数据  
df = pd.read_excel('/Users/macbook/Documents/data/FL200_N.xlsx', engine='openpyxl')  
data = df['N%']  
  
# 计算箱子的宽度  
bin_width = 1  # 设定箱子宽度  
  
# 修改数据范围以适应12个箱子  
data_min = 2  
data_max = 4.5  
bin_width = (data_max - data_min) / 12  # 计算每个箱子的宽度  
bin_edges = np.arange(data_min, data_max + bin_width, bin_width)  # 创建箱子边界  
bins = bin_edges[:-1]  # 移除最后一个箱子边界，因为最后一个箱子到最大值包括最大值本身 
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2  # 计算箱子中心  
  
# 创建直方图，使用自定义的箱子中心作为x轴的值  
plt.hist(data, bins=bin_centers, edgecolor='black', histtype='bar', label='Frequency', rwidth=0.8)  
  
# 创建一个新的x轴，并将刻度位置设为每个箱子的中心  
new_xaxis = plt.twinx()  
new_xaxis.spines['bottom'].set_position('zero')  # 将原来的x轴移至y=0处  
new_xaxis.set_xticks(bin_centers)  # 设置新的x轴刻度位置  
new_xaxis.set_xticklabels(bin_centers, rotation=45)  # 设置新的x轴刻度标签为每个箱子的中心值  
  
# 删除原来的x轴和y轴的标签  
plt.xticks([], [])  
plt.yticks([], [])  
  
# 设置图表标题和坐标轴标签  
plt.title('Distribution of Column1')  
plt.xlabel('Column1')  
plt.ylabel('Frequency')  
  
# 显示图例，并将图例放在右下角  
plt.legend(loc='upper right')  
  
# 显示直方图和新的x轴  
plt.show()
```
