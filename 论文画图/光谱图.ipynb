```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
import pandas as pd
# 设置显示中文字体，
from pylab import mpl
import math
# 设置font.sans-serif 或 font.family 均可
mpl.rcParams["font.sans-serif"] = ["Arial Unicode MS"]  ## mac
plt.rcParams['font.family'] = ['Arial Unicode MS']  ## mac
# 设置正常显示符号
mpl.rcParams["axes.unicode_minus"] = False
import warnings
warnings.filterwarnings(action='ignore')  # 忽略告警

# 读取数据
data = pd.read_excel("/Users/macbook/folder/data/虚拟样本/FL199.xlsx")
X = data.values[:, 1:]
y = data.values[:, 0]
print(X.shape)

# 绘制原始光谱图
plt.figure(figsize=(7, 6))
plt.plot(np.linspace(942, 1701, X.shape[1]), np.transpose(X))
fonts = 20
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.xlabel('波长(nm)', fontsize=fonts)
plt.ylabel('反射率', fontsize=fonts)
plt.tight_layout()
plt.grid(False)

# 在固定位置添加 "(a)"
plt.text(920, 0.61, "(RAW)", fontsize=16, color='black')
plt.gcf().tight_layout(pad=0,rect=(0.01,0.01,0.95,0.98))
# 保存为TIFF格式
plt.savefig('/Users/macbook/Desktop/大论文/a.tif', format='tiff', dpi=500, bbox_inches='tight')
# 保存为JPG格式
plt.savefig('/Users/macbook/Desktop/大论文/a.jpg', format='jpg', dpi=500, bbox_inches='tight')
# 保存为SVG格式
# plt.savefig('/Users/macbook/Desktop/论文11_24/图2d.svg')
# plt.savefig('/Users/macbook/Desktop/论文11_24/图2d.eps',format='eps')
# plt.savefig('/Users/macbook/Desktop/论文11_24/图2d.pdf',format='pdf')
plt.show() 
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
import pandas as pd
from sklearn.model_selection import cross_val_score

# 读取数据
data = pd.read_excel("/Users/macbook/folder/data/虚拟样本/smote_FL199.xlsx")
X = data.values[:, 1:]
y = data.values[:, 0]
print(X.shape)

# 绘制原始光谱图
plt.figure(figsize=(7, 6))
plt.plot(np.linspace(942, 1701, X.shape[1]), np.transpose(X))
fonts = 20
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

# Customize y-axis tick intervals and decimal places
plt.gca().get_yaxis().set_major_locator(plt.MaxNLocator(integer=False))  # To allow non-integer tick intervals
plt.gca().yaxis.set_major_formatter(plt.FormatStrFormatter('%.2f'))  # Format tick labels to two decimal places

plt.xlabel('波长(nm)', fontsize=fonts)
plt.ylabel('反射率', fontsize=fonts)
plt.tight_layout()
plt.grid(False)

# 在最左上角添加 "(a)"
plt.text(915, max(np.max(X, axis=0))*1.005 , "(SMOTE)", fontsize=16, color='black', va='top', ha='left')
plt.gcf().tight_layout(pad=0,rect=(0.01,0.01,0.95,0.98))
# 保存为TIFF格式
plt.savefig('/Users/macbook/Desktop/大论文/a2.tif', format='tiff', dpi=300, bbox_inches='tight')
# 保存为JPG格式
plt.savefig('/Users/macbook/Desktop/大论文/a2.jpg', format='jpg', dpi=300, bbox_inches='tight')

# plt.savefig('/Users/macbook/Desktop/论文11_24/图2a.svg')
# plt.savefig('/Users/macbook/Desktop/论文11_24/图2a.eps',format='eps')
# plt.savefig('/Users/macbook/Desktop/论文11_24/图2a.pdf',format='pdf')
plt.show() 
```
